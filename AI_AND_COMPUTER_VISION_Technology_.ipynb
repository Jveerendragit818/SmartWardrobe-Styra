```python

# Task 3: Test a pretrained clothing model

from google.colab import files
from PIL import Image
from transformers import pipeline
import os

# Upload 3–5 JPG or PNG clothing images
uploaded = files.upload()

# Load the pretrained model (first run downloads the weights)
print("Loading FashionCLIP. This may take a few minutes...")
classifier = pipeline(
    "zero-shot-image-classification",
    model="patrickjohncyh/fashion-clip",
    device=0 if __import__("torch").cuda.is_available() else -1
)

categories = [
    "a photograph of a t-shirt",
    "a photograph of a shirt",
    "a photograph of jeans",
    "a photograph of a dress",
    "a photograph of a jacket",
    "a photograph of a hoodie",
    "a photograph of a skirt",
    "a photograph of trousers"
]

colours = [
    "a red garment", "a blue garment", "a black garment",
    "a white garment", "a green garment", "a yellow garment",
    "a pink garment", "a brown garment"
]

patterns = [
    "a plain solid-colour garment",
    "a striped garment",
    "a checked garment",
    "a floral-print garment",
    "a polka-dot garment",
    "a graphic-print garment"
]

results = []

for filename in uploaded:
    try:
        image = Image.open(filename).convert("RGB")
        print(f"\nImage: {filename} | Size: {image.size}")

        for attribute, labels in [
            ("Category", categories),
            ("Colour", colours),
            ("Pattern", patterns)
        ]:
            predictions = classifier(
                image,
                candidate_labels=labels
            )
            best = predictions[0]
            print(
                f"{attribute}: {best['label']} "
                f"(similarity score: {best['score']:.3f})"
            )
            results.append({
                "image": filename,
                "attribute": attribute,
                "prediction": best["label"],
                "score": round(best["score"], 3)
            })

    except Exception as error:
        print("Could not process", filename, ":", error)

import pandas as pd

df = pd.DataFrame(results)
display(df)

df.to_csv("/content/clothing_ai/results/test_results.csv", index=False)
print("Results saved to test_results.csv")
```
